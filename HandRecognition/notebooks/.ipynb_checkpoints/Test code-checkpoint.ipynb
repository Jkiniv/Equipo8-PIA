```python
import cv2
import mediapipe as mp

from mediapipe.tasks import python
from mediapipe.tasks.python import vision

MODEL_PATH = r"C:\Proyectos\LSM_Proyecto\modelos\hand_landmarker.task"

base_options = python.BaseOptions(
    model_asset_path=MODEL_PATH
)

options = vision.HandLandmarkerOptions(
    base_options=base_options,
    num_hands=2
)

detector = vision.HandLandmarker.create_from_options(
    options
)

cap = cv2.VideoCapture(0)

while True:

    success, frame = cap.read()

    if not success:
        break

    rgb = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2RGB
    )

    mp_image = mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=rgb
    )

    result = detector.detect(mp_image)

    h, w, _ = frame.shape

    for hand in result.hand_landmarks:

        for landmark in hand:

            x = int(landmark.x * w)
            y = int(landmark.y * h)

            cv2.circle(
                frame,
                (x, y),
                5,
                (0, 255, 0),
                -1
            )

    cv2.imshow(
        "HandLandmarker Test",
        frame
    )

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import mediapipe as mp

from mediapipe.tasks import python
from mediapipe.tasks.python import vision

MODEL_PATH = r"C:\Proyectos\LSM_Proyecto\modelos\hand_landmarker.task"

base_options = python.BaseOptions(
    model_asset_path=MODEL_PATH
)

options = vision.HandLandmarkerOptions(
    base_options=base_options,
    num_hands=1
)

detector = vision.HandLandmarker.create_from_options(
    options
)

cap = cv2.VideoCapture(0)

while True:

    success, frame = cap.read()

    if not success:
        break

    rgb = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2RGB
    )

    mp_image = mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=rgb
    )

    result = detector.detect(mp_image)

    if result.hand_landmarks:

        hand = result.hand_landmarks[0]

        print("\n" + "=" * 60)

        for i, landmark in enumerate(hand):

            print(
                f"Punto {i:02d} | "
                f"x={landmark.x:.4f} | "
                f"y={landmark.y:.4f} | "
                f"z={landmark.z:.4f}"
            )

    cv2.imshow(
        "Landmarks Coordinates",
        frame
    )

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import mediapipe as mp

from mediapipe.tasks import python
from mediapipe.tasks.python import vision

MODEL_PATH = r"C:\Proyectos\LSM_Proyecto\modelos\hand_landmarker.task"

base_options = python.BaseOptions(
    model_asset_path=MODEL_PATH
)

options = vision.HandLandmarkerOptions(
    base_options=base_options,
    num_hands=2
)

detector = vision.HandLandmarker.create_from_options(options)

cap = cv2.VideoCapture(0)

while True:

    success, frame = cap.read()

    if not success:
        break

    rgb = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2RGB
    )

    mp_image = mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=rgb
    )

    result = detector.detect(mp_image)

    h, w, _ = frame.shape

    if result.hand_landmarks:

        nearest_hand = min(
            result.hand_landmarks,
            key=lambda hand:
            sum(point.z for point in hand) / len(hand)
        )

        for landmark in nearest_hand:

            x = int(landmark.x * w)
            y = int(landmark.y * h)

            cv2.circle(
                frame,
                (x, y),
                5,
                (0, 255, 0),
                -1
            )

    cv2.imshow(
        "LSM - Mano Mas Cercana",
        frame
    )

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import csv
import os
import mediapipe as mp
import time

from mediapipe.tasks import python
from mediapipe.tasks.python import vision

# ===================================
# CONFIGURACION
# ===================================

# Cambia esta letra antes de ejecutar
LETRA = "A"

# Cantidad total de muestras
MAX_MUESTRAS = 500

# Tiempo entre cada muestra
INTERVALO_CAPTURA = 0.10

MODEL_PATH = r"C:\Proyectos\LSM_Proyecto\modelos\hand_landmarker.task"

CSV_PATH = rf"C:\Proyectos\LSM_Proyecto\dataset\{LETRA}.csv"

# ===================================

# Crea la carpeta dataset si no existe
os.makedirs(
    os.path.dirname(CSV_PATH),
    exist_ok=True
)

base_options = python.BaseOptions(
    model_asset_path=MODEL_PATH
)

options = vision.HandLandmarkerOptions(
    base_options=base_options,
    num_hands=2
)

detector = vision.HandLandmarker.create_from_options(
    options
)

cap = cv2.VideoCapture(0)

contador = 0
capturando = False

with open(
    CSV_PATH,
    "w",
    newline=""
) as file:

    writer = csv.writer(file)

    while contador < MAX_MUESTRAS:

        success, frame = cap.read()

        if not success:
            print("No se pudo obtener imagen de la cámara.")
            break

        # Lee las teclas presionadas
        key = cv2.waitKey(1) & 0xFF

        # ===================================
        # ESPACIO = INICIAR O PAUSAR
        # ===================================

        if key == 32:

            if not capturando:

                print("Captura iniciando en 3 segundos...")

                cv2.putText(
                    frame,
                    "Iniciando en 3 segundos...",
                    (20, 100),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.8,
                    (0, 255, 255),
                    2
                )

                cv2.imshow(
                    "Capturando Dataset",
                    frame
                )

                cv2.waitKey(1)

                time.sleep(3)

                capturando = True

                print("Capturando muestras...")

            else:

                capturando = False

                print("Captura pausada.")

        # ===================================
        # PROCESAMIENTO DE LA IMAGEN
        # ===================================

        rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        mp_image = mp.Image(
            image_format=mp.ImageFormat.SRGB,
            data=rgb
        )

        result = detector.detect(mp_image)

        # ===================================
        # TEXTO DEL ESTADO
        # ===================================

        if not capturando:

            cv2.putText(
                frame,
                "PAUSADO - Presiona ESPACIO para iniciar",
                (20, 50),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 255),
                2
            )

        else:

            cv2.putText(
                frame,
                f"CAPTURANDO {LETRA}: {contador}/{MAX_MUESTRAS}",
                (20, 50),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (0, 255, 0),
                2
            )

        # ===================================
        # SELECCIONAR LA MANO MAS CERCANA
        # ===================================

        if result.hand_landmarks and capturando:

            nearest_hand = min(
                result.hand_landmarks,
                key=lambda hand:
                sum(point.z for point in hand) / len(hand)
            )

            row = []

            for landmark in nearest_hand:

                row.extend([
                    landmark.x,
                    landmark.y,
                    landmark.z
                ])

            # Agrega la etiqueta al final de la fila
            row.append(LETRA)

            # Guarda la muestra en el CSV
            writer.writerow(row)

            contador += 1

            print(
                f"{LETRA}: {contador}/{MAX_MUESTRAS}"
            )

            time.sleep(INTERVALO_CAPTURA)

        # ===================================
        # MOSTRAR LA CÁMARA
        # ===================================

        cv2.imshow(
            "Capturando Dataset",
            frame
        )

        # ESC = cerrar antes de terminar
        if key == 27:
            print("Captura detenida manualmente.")
            break

cap.release()
detector.close()
cv2.destroyAllWindows()

print(
    f"\nDataset guardado en:\n{CSV_PATH}"
)

print(
    f"Muestras guardadas: {contador}"
)
```
