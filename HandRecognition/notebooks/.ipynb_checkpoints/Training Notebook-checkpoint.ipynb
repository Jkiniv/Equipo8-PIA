```python
import os
import time
import joblib
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# ==========================================================
# CONFIGURACION
# ==========================================================

DATASET_DIR = r"C:\Proyectos\LSM_Proyecto\dataset"

MODELOS_DIR = r"C:\Proyectos\LSM_Proyecto\modelos"

# Letras que se utilizaran en esta primera prueba
LETRAS = [
    "A",
    "B",
    "C",
    "D",
    "E",
    "F",
    "G",
    "H",
    "I",
    "J"
]

FRAMES_POR_MUESTRA = 20
LANDMARKS_POR_FRAME = 21
COORDENADAS_POR_LANDMARK = 3

VARIABLES_POR_MUESTRA = (
    FRAMES_POR_MUESTRA
    * LANDMARKS_POR_FRAME
    * COORDENADAS_POR_LANDMARK
)

# Porcentaje final de cada CSV usado para prueba
PORCENTAJE_PRUEBA = 0.20

RANDOM_STATE = 42

# ==========================================================
# CREAR CARPETA DE MODELOS
# ==========================================================

os.makedirs(
    MODELOS_DIR,
    exist_ok=True
)

# ==========================================================
# NORMALIZACION HIBRIDA
# ==========================================================

def normalizar_secuencia(secuencia_plana):
    """
    Recibe una muestra con 1260 valores:

    20 frames
    x 21 landmarks
    x 3 coordenadas

    Devuelve otros 1260 valores organizados como:

    - Trayectoria de la muneca respecto al primer frame.
    - Landmarks 1 a 20 relativos a la muneca.
    - Coordenadas divididas por el tamano de la mano.

    Esto elimina gran parte de la dependencia respecto a:
    - Posicion absoluta en la pantalla.
    - Cercania o distancia a la camara.

    Pero conserva:
    - Movimiento global de la muneca.
    - Forma relativa de los dedos.
    - Movimiento interno de los dedos.
    """

    secuencia = np.asarray(
        secuencia_plana,
        dtype=np.float32
    ).reshape(
        FRAMES_POR_MUESTRA,
        LANDMARKS_POR_FRAME,
        COORDENADAS_POR_LANDMARK
    )

    # Landmark 0 = muneca
    munecas = secuencia[:, 0, :]

    # Landmark 9 = base del dedo medio
    base_dedo_medio = secuencia[:, 9, :]

    # Distancia muneca - base del dedo medio
    escalas_por_frame = np.linalg.norm(
        base_dedo_medio - munecas,
        axis=1
    )

    # Usar la escala mediana de toda la secuencia
    # para evitar cambios bruscos frame por frame
    escala = np.median(
        escalas_por_frame
    )

    # Evitar divisiones entre cero
    if escala < 1e-6:
        escala = 1.0

    # Posicion inicial de la muneca
    muneca_inicial = munecas[0].copy()

    secuencia_normalizada = []

    for frame_index in range(FRAMES_POR_MUESTRA):

        frame = secuencia[frame_index]

        muneca_actual = frame[0]

        # ------------------------------------------
        # TRAYECTORIA GLOBAL DE LA MUNECA
        # ------------------------------------------

        trayectoria_muneca = (
            muneca_actual - muneca_inicial
        ) / escala

        datos_frame = [
            trayectoria_muneca[0],
            trayectoria_muneca[1],
            trayectoria_muneca[2]
        ]

        # ------------------------------------------
        # LANDMARKS RELATIVOS A LA MUNECA
        # ------------------------------------------

        for landmark_index in range(
            1,
            LANDMARKS_POR_FRAME
        ):

            landmark_relativo = (
                frame[landmark_index]
                - muneca_actual
            ) / escala

            datos_frame.extend([
                landmark_relativo[0],
                landmark_relativo[1],
                landmark_relativo[2]
            ])

        # Cada frame sigue teniendo:
        # 3 valores de trayectoria de muneca
        # + 20 landmarks x 3 coordenadas
        # = 63 valores

        secuencia_normalizada.extend(
            datos_frame
        )

    return np.asarray(
        secuencia_normalizada,
        dtype=np.float32
    )

# ==========================================================
# CARGAR DATASETS
# ==========================================================

def cargar_datasets():

    X_train_total = []
    y_train_total = []

    X_test_total = []
    y_test_total = []

    resumen = []

    print("=" * 70)
    print("CARGANDO DATASETS")
    print("=" * 70)

    for letra in LETRAS:

        ruta_csv = os.path.join(
            DATASET_DIR,
            f"{letra}.csv"
        )

        if not os.path.exists(ruta_csv):

            print(
                f"ERROR: No se encontro el archivo:"
            )
            print(ruta_csv)

            raise FileNotFoundError(
                ruta_csv
            )

        # Los CSV no tienen encabezados
        df = pd.read_csv(
            ruta_csv,
            header=None
        )

        columnas_encontradas = df.shape[1]

        columnas_esperadas = (
            VARIABLES_POR_MUESTRA + 1
        )

        if columnas_encontradas != columnas_esperadas:

            print(
                f"\nERROR EN {letra}.csv"
            )

            print(
                f"Columnas encontradas: "
                f"{columnas_encontradas}"
            )

            print(
                f"Columnas esperadas: "
                f"{columnas_esperadas}"
            )

            print(
                "Cada fila debe contener "
                "1260 coordenadas y una letra."
            )

            raise ValueError(
                f"Estructura incorrecta en "
                f"{letra}.csv"
            )

        # Coordenadas
        X_letra = df.iloc[
            :,
            :VARIABLES_POR_MUESTRA
        ].to_numpy(
            dtype=np.float32
        )

        # Etiquetas
        y_letra = df.iloc[
            :,
            VARIABLES_POR_MUESTRA
        ].astype(str).to_numpy()

        # Verificar valores invalidos
        if np.isnan(X_letra).any():

            raise ValueError(
                f"{letra}.csv contiene valores NaN."
            )

        if np.isinf(X_letra).any():

            raise ValueError(
                f"{letra}.csv contiene valores infinitos."
            )

        # Verificar etiqueta
        etiquetas_encontradas = set(
            y_letra
        )

        if etiquetas_encontradas != {letra}:

            print(
                f"ERROR: {letra}.csv contiene "
                f"estas etiquetas:"
            )

            print(
                etiquetas_encontradas
            )

            raise ValueError(
                f"Etiquetas incorrectas en "
                f"{letra}.csv"
            )

        print(
            f"Normalizando letra {letra}..."
        )

        X_normalizado = np.asarray([
            normalizar_secuencia(muestra)
            for muestra in X_letra
        ])

        cantidad_total = len(
            X_normalizado
        )

        cantidad_test = max(
            1,
            int(
                cantidad_total
                * PORCENTAJE_PRUEBA
            )
        )

        cantidad_train = (
            cantidad_total
            - cantidad_test
        )

        # Se usa la parte final de cada archivo
        # como prueba.
        #
        # Esto evita mezclar completamente
        # secuencias consecutivas casi identicas.

        X_train_letra = (
            X_normalizado[
                :cantidad_train
            ]
        )

        y_train_letra = (
            y_letra[
                :cantidad_train
            ]
        )

        X_test_letra = (
            X_normalizado[
                cantidad_train:
            ]
        )

        y_test_letra = (
            y_letra[
                cantidad_train:
            ]
        )

        X_train_total.append(
            X_train_letra
        )

        y_train_total.append(
            y_train_letra
        )

        X_test_total.append(
            X_test_letra
        )

        y_test_total.append(
            y_test_letra
        )

        resumen.append({
            "Letra": letra,
            "Total": cantidad_total,
            "Entrenamiento": cantidad_train,
            "Prueba": cantidad_test
        })

        print(
            f"{letra}: "
            f"{cantidad_total} muestras | "
            f"Train: {cantidad_train} | "
            f"Test: {cantidad_test}"
        )

    X_train = np.vstack(
        X_train_total
    )

    y_train = np.concatenate(
        y_train_total
    )

    X_test = np.vstack(
        X_test_total
    )

    y_test = np.concatenate(
        y_test_total
    )

    # Revolver solamente el conjunto
    # de entrenamiento.
    generador = np.random.default_rng(
        RANDOM_STATE
    )

    indices_train = generador.permutation(
        len(X_train)
    )

    X_train = X_train[
        indices_train
    ]

    y_train = y_train[
        indices_train
    ]

    # Revolver prueba sin mezclarla
    # nuevamente con entrenamiento
    indices_test = generador.permutation(
        len(X_test)
    )

    X_test = X_test[
        indices_test
    ]

    y_test = y_test[
        indices_test
    ]

    print("\n" + "=" * 70)
    print("RESUMEN")
    print("=" * 70)

    resumen_df = pd.DataFrame(
        resumen
    )

    print(
        resumen_df.to_string(
            index=False
        )
    )

    print(
        f"\nEntrenamiento total: "
        f"{len(X_train)}"
    )

    print(
        f"Prueba total: "
        f"{len(X_test)}"
    )

    print(
        f"Variables por muestra: "
        f"{X_train.shape[1]}"
    )

    return (
        X_train,
        X_test,
        y_train,
        y_test
    )

# ==========================================================
# FUNCION DE EVALUACION
# ==========================================================

def evaluar_modelo(
    nombre,
    modelo,
    X_test,
    y_test
):

    predicciones = modelo.predict(
        X_test
    )

    precision = accuracy_score(
        y_test,
        predicciones
    )

    print("\n" + "=" * 70)
    print(f"RESULTADOS: {nombre}")
    print("=" * 70)

    print(
        f"Accuracy: "
        f"{precision * 100:.2f}%"
    )

    print("\nReporte de clasificacion:")

    print(
        classification_report(
            y_test,
            predicciones,
            labels=LETRAS,
            zero_division=0
        )
    )

    matriz = confusion_matrix(
        y_test,
        predicciones,
        labels=LETRAS
    )

    matriz_df = pd.DataFrame(
        matriz,
        index=[
            f"Real_{letra}"
            for letra in LETRAS
        ],
        columns=[
            f"Pred_{letra}"
            for letra in LETRAS
        ]
    )

    print("Matriz de confusion:")

    print(
        matriz_df.to_string()
    )

    ruta_matriz = os.path.join(
        MODELOS_DIR,
        f"matriz_confusion_{nombre}.csv"
    )

    matriz_df.to_csv(
        ruta_matriz
    )

    print(
        f"\nMatriz guardada en:"
    )

    print(
        ruta_matriz
    )

    return precision

# ==========================================================
# PROGRAMA PRINCIPAL
# ==========================================================

print("\n" + "=" * 70)
print("ENTRENAMIENTO DEL RECONOCEDOR LSM")
print("=" * 70)

X_train, X_test, y_train, y_test = (
    cargar_datasets()
)

# ==========================================================
# MODELO 1: RANDOM FOREST
# ==========================================================

print("\n" + "=" * 70)
print("ENTRENANDO RANDOM FOREST")
print("=" * 70)

inicio_rf = time.perf_counter()

modelo_rf = RandomForestClassifier(
    n_estimators=400,
    max_features="sqrt",
    min_samples_leaf=2,
    class_weight="balanced",
    n_jobs=-1,
    random_state=RANDOM_STATE,
    verbose=1
)

modelo_rf.fit(
    X_train,
    y_train
)

tiempo_rf = (
    time.perf_counter()
    - inicio_rf
)

precision_rf = evaluar_modelo(
    "random_forest",
    modelo_rf,
    X_test,
    y_test
)

ruta_rf = os.path.join(
    MODELOS_DIR,
    "modelo_random_forest.joblib"
)

joblib.dump(
    modelo_rf,
    ruta_rf
)

print(
    f"\nRandom Forest guardado en:\n"
    f"{ruta_rf}"
)

print(
    f"Tiempo de entrenamiento: "
    f"{tiempo_rf:.2f} segundos"
)

# ==========================================================
# MODELO 2: RED NEURONAL MLP
# ==========================================================

print("\n" + "=" * 70)
print("ENTRENANDO RED NEURONAL MLP")
print("=" * 70)

inicio_mlp = time.perf_counter()

# El StandardScaler queda guardado dentro
# del Pipeline junto con la red neuronal.
modelo_mlp = Pipeline([
    (
        "escalador",
        StandardScaler()
    ),
    (
        "red_neuronal",
        MLPClassifier(
            hidden_layer_sizes=(
                256,
                128,
                64
            ),
            activation="relu",
            solver="adam",
            alpha=0.001,
            batch_size=64,
            learning_rate="adaptive",
            learning_rate_init=0.001,
            max_iter=500,
            early_stopping=True,
            validation_fraction=0.15,
            n_iter_no_change=25,
            random_state=RANDOM_STATE,
            verbose=True
        )
    )
])

modelo_mlp.fit(
    X_train,
    y_train
)

tiempo_mlp = (
    time.perf_counter()
    - inicio_mlp
)

precision_mlp = evaluar_modelo(
    "mlp",
    modelo_mlp,
    X_test,
    y_test
)

ruta_mlp = os.path.join(
    MODELOS_DIR,
    "modelo_mlp.joblib"
)

joblib.dump(
    modelo_mlp,
    ruta_mlp
)

print(
    f"\nRed neuronal MLP guardada en:\n"
    f"{ruta_mlp}"
)

print(
    f"Tiempo de entrenamiento: "
    f"{tiempo_mlp:.2f} segundos"
)

# ==========================================================
# COMPARACION
# ==========================================================

print("\n" + "=" * 70)
print("COMPARACION FINAL")
print("=" * 70)

print(
    f"Random Forest: "
    f"{precision_rf * 100:.2f}%"
)

print(
    f"Red neuronal MLP: "
    f"{precision_mlp * 100:.2f}%"
)

if precision_rf > precision_mlp:

    mejor_modelo = modelo_rf
    mejor_nombre = "Random Forest"
    mejor_precision = precision_rf

elif precision_mlp > precision_rf:

    mejor_modelo = modelo_mlp
    mejor_nombre = "Red neuronal MLP"
    mejor_precision = precision_mlp

else:

    mejor_modelo = modelo_mlp
    mejor_nombre = "Red neuronal MLP"
    mejor_precision = precision_mlp

ruta_mejor = os.path.join(
    MODELOS_DIR,
    "mejor_modelo_lsm.joblib"
)

joblib.dump(
    mejor_modelo,
    ruta_mejor
)

print(
    f"\nMejor modelo: "
    f"{mejor_nombre}"
)

print(
    f"Accuracy: "
    f"{mejor_precision * 100:.2f}%"
)

print(
    f"Modelo guardado en:\n"
    f"{ruta_mejor}"
)

# Guardar configuracion necesaria
# para el reconocimiento en tiempo real
configuracion = {
    "letras": LETRAS,
    "frames_por_muestra": FRAMES_POR_MUESTRA,
    "landmarks_por_frame": LANDMARKS_POR_FRAME,
    "coordenadas_por_landmark":
        COORDENADAS_POR_LANDMARK,
    "variables_por_muestra":
        VARIABLES_POR_MUESTRA,
    "tipo_normalizacion":
        "hibrida_muneca_trayectoria"
}

ruta_configuracion = os.path.join(
    MODELOS_DIR,
    "configuracion_modelo.joblib"
)

joblib.dump(
    configuracion,
    ruta_configuracion
)

print(
    f"\nConfiguracion guardada en:\n"
    f"{ruta_configuracion}"
)

print("\nEntrenamiento terminado.")
```
