```python
import cv2
import csv
import os
import mediapipe as mp
import time

from mediapipe.tasks import python
from mediapipe.tasks.python import vision

# ===================================
# CONFIGURACION
# ===================================

# Cambia esta letra antes de ejecutar
LETRA = "A"

# Cantidad de secuencias completas
MAX_MUESTRAS = 300

# Cada muestra contiene 20 frames
FRAMES_POR_MUESTRA = 20

# 0.05 segundos = 20 frames efectivos por segundo
INTERVALO_FRAME = 0.05

MODEL_PATH = (
    r"C:\Proyectos\LSM_Proyecto"
    r"\modelos\hand_landmarker.task"
)

CSV_PATH = rf"C:\Proyectos\LSM_Proyecto\dataset\{LETRA}.csv"

# ===================================
# PREPARACION
# ===================================

os.makedirs(
    os.path.dirname(CSV_PATH),
    exist_ok=True
)

base_options = python.BaseOptions(
    model_asset_path=MODEL_PATH
)

options = vision.HandLandmarkerOptions(
    base_options=base_options,
    num_hands=2
)

detector = vision.HandLandmarker.create_from_options(
    options
)

cap = cv2.VideoCapture(0)

# ===================================
# VARIABLES DE CONTROL
# ===================================

contador_muestras = 0
capturando = False

# Aquí se almacenan temporalmente los 1260 valores
secuencia_actual = []

# Contador de frames de la muestra actual
contador_frames = 0

# Controla el intervalo de 0.05 segundos
ultimo_tiempo_captura = 0.0

# ===================================
# CREAR CSV
# ===================================

with open(
    CSV_PATH,
    "w",
    newline=""
) as file:

    writer = csv.writer(file)

    while contador_muestras < MAX_MUESTRAS:

        success, frame = cap.read()

        if not success:
            print("No se pudo obtener una imagen de la cámara.")
            break

        # Reflejar la cámara como espejo
        frame = cv2.flip(frame, 1)

        key = cv2.waitKey(1) & 0xFF

        # ===================================
        # ESPACIO = INICIAR O PAUSAR
        # ===================================

        if key == 32:

            if not capturando:

                print("Captura iniciando en 3 segundos...")

                cv2.putText(
                    frame,
                    "Iniciando en 3 segundos...",
                    (20, 100),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.8,
                    (0, 255, 255),
                    2
                )

                cv2.imshow(
                    "Capturando Dataset Temporal",
                    frame
                )

                cv2.waitKey(1)

                time.sleep(3)

                # Limpiar cualquier secuencia incompleta
                secuencia_actual = []
                contador_frames = 0

                # Iniciar el reloj de captura
                ultimo_tiempo_captura = time.perf_counter()

                capturando = True

                print("Capturando secuencias...")

            else:

                capturando = False

                # Se elimina la secuencia parcial
                secuencia_actual = []
                contador_frames = 0

                print(
                    "Captura pausada. "
                    "La secuencia incompleta fue descartada."
                )

        # ===================================
        # PROCESAR FRAME CON MEDIAPIPE
        # ===================================

        rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        mp_image = mp.Image(
            image_format=mp.ImageFormat.SRGB,
            data=rgb
        )

        result = detector.detect(mp_image)

        mano_detectada = False
        nearest_hand = None

        if result.hand_landmarks:

            mano_detectada = True

            # Seleccionar únicamente la mano más cercana
            nearest_hand = min(
                result.hand_landmarks,
                key=lambda hand: (
                    sum(point.z for point in hand)
                    / len(hand)
                )
            )

            # Dibujar los 21 puntos de la mano seleccionada
            alto, ancho, _ = frame.shape

            for landmark in nearest_hand:

                x = int(landmark.x * ancho)
                y = int(landmark.y * alto)

                cv2.circle(
                    frame,
                    (x, y),
                    4,
                    (255, 0, 255),
                    -1
                )

        # ===================================
        # CAPTURAR UN FRAME CADA 0.05 SEGUNDOS
        # ===================================

        tiempo_actual = time.perf_counter()

        if (
            capturando
            and mano_detectada
            and nearest_hand is not None
            and tiempo_actual - ultimo_tiempo_captura
            >= INTERVALO_FRAME
        ):

            datos_frame = []

            # Guardar X, Y y Z de los 21 puntos
            for landmark in nearest_hand:

                datos_frame.extend([
                    landmark.x,
                    landmark.y,
                    landmark.z
                ])

            # Agregar los 63 valores a la secuencia actual
            secuencia_actual.extend(datos_frame)

            contador_frames += 1

            ultimo_tiempo_captura = tiempo_actual

            print(
                f"Muestra {contador_muestras + 1}/"
                f"{MAX_MUESTRAS} | "
                f"Frame {contador_frames}/"
                f"{FRAMES_POR_MUESTRA}"
            )

            # ===================================
            # GUARDAR SECUENCIA COMPLETA
            # ===================================

            if contador_frames == FRAMES_POR_MUESTRA:

                # Agregar la etiqueta al final
                fila = secuencia_actual + [LETRA]

                # La fila tiene 1260 valores + etiqueta
                writer.writerow(fila)

                # Asegurar escritura inmediata
                file.flush()

                contador_muestras += 1

                print(
                    f"Secuencia completa guardada: "
                    f"{contador_muestras}/"
                    f"{MAX_MUESTRAS}"
                )

                # Reiniciar para la siguiente muestra
                secuencia_actual = []
                contador_frames = 0

        # ===================================
        # MENSAJES EN PANTALLA
        # ===================================

        if not capturando:

            cv2.putText(
                frame,
                "PAUSADO - ESPACIO para iniciar",
                (20, 40),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 255),
                2
            )

        else:

            cv2.putText(
                frame,
                f"CAPTURANDO LETRA {LETRA}",
                (20, 40),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (0, 255, 0),
                2
            )

            cv2.putText(
                frame,
                (
                    f"Muestra: "
                    f"{contador_muestras}/"
                    f"{MAX_MUESTRAS}"
                ),
                (20, 75),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 0),
                2
            )

            cv2.putText(
                frame,
                (
                    f"Frame: "
                    f"{contador_frames}/"
                    f"{FRAMES_POR_MUESTRA}"
                ),
                (20, 110),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 0),
                2
            )

        if not mano_detectada:

            cv2.putText(
                frame,
                "MANO NO DETECTADA",
                (20, 150),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 0, 255),
                2
            )

        cv2.imshow(
            "Capturando Dataset Temporal",
            frame
        )

        # ESC = terminar manualmente
        if key == 27:

            print("Captura detenida manualmente.")

            break

# ===================================
# CERRAR RECURSOS
# ===================================

cap.release()
detector.close()
cv2.destroyAllWindows()

print("\nCaptura finalizada.")

print(
    f"Dataset guardado en:\n{CSV_PATH}"
)

print(
    f"Secuencias completas guardadas: "
    f"{contador_muestras}"
)

print(
    "Variables por secuencia: "
    f"{FRAMES_POR_MUESTRA * 21 * 3}"
)
```

```python
import cv2
import csv
import os
import mediapipe as mp
import time

from mediapipe.tasks import python
from mediapipe.tasks.python import vision

# ===================================
# CONFIGURACION
# ===================================

# Cambia esta letra antes de ejecutar
LETRA = "C"

# Cantidad de secuencias completas
MAX_MUESTRAS = 300

# Cada muestra contiene 20 frames
FRAMES_POR_MUESTRA = 20

# 0.05 segundos entre frames
INTERVALO_FRAME = 0.05

# Tiempo para cambiar de posicion entre muestras
TIEMPO_ACOMODO = 1.0

MODEL_PATH = (
    r"C:\Proyectos\LSM_Proyecto"
    r"\modelos\hand_landmarker.task"
)

CSV_PATH = rf"C:\Proyectos\LSM_Proyecto\dataset\{LETRA}.csv"

# ===================================
# PREPARACION
# ===================================

os.makedirs(
    os.path.dirname(CSV_PATH),
    exist_ok=True
)

base_options = python.BaseOptions(
    model_asset_path=MODEL_PATH
)

options = vision.HandLandmarkerOptions(
    base_options=base_options,
    num_hands=2
)

detector = vision.HandLandmarker.create_from_options(
    options
)

cap = cv2.VideoCapture(0)

# ===================================
# VARIABLES DE CONTROL
# ===================================

contador_muestras = 0
contador_frames = 0

# El programa inicia pausado
programa_activo = False

# Estados posibles:
# "ACOMODANDO"
# "CAPTURANDO"
estado = "ACOMODANDO"

secuencia_actual = []

# Guarda la lateralidad detectada en cada frame
lateralidades_secuencia = []

inicio_acomodo = 0.0
ultimo_tiempo_captura = 0.0

# ===================================
# CREAR CSV
# ===================================

with open(
    CSV_PATH,
    "w",
    newline=""
) as file:

    writer = csv.writer(file)

    while contador_muestras < MAX_MUESTRAS:

        success, frame = cap.read()

        if not success:
            print("No se pudo obtener imagen de la camara.")
            break

        # Vista tipo espejo
        frame = cv2.flip(frame, 1)

        key = cv2.waitKey(1) & 0xFF

        # ===================================
        # ESPACIO = INICIAR O PAUSAR
        # ===================================

        if key == 32:

            if not programa_activo:

                programa_activo = True
                estado = "ACOMODANDO"

                secuencia_actual = []
                lateralidades_secuencia = []
                contador_frames = 0

                inicio_acomodo = time.perf_counter()

                print("Programa iniciado.")
                print("Tienes 1 segundo para acomodarte.")

            else:

                programa_activo = False
                estado = "ACOMODANDO"

                # Descartar muestra incompleta
                secuencia_actual = []
                lateralidades_secuencia = []
                contador_frames = 0

                print("Captura pausada.")
                print("La muestra incompleta fue descartada.")

        # ===================================
        # PROCESAR LA IMAGEN
        # ===================================

        rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        mp_image = mp.Image(
            image_format=mp.ImageFormat.SRGB,
            data=rgb
        )

        result = detector.detect(mp_image)

        mano_detectada = False
        nearest_hand = None
        lateralidad_actual = "Unknown"

        if result.hand_landmarks:

            mano_detectada = True

            # ===================================
            # SELECCIONAR LA MANO MAS CERCANA
            # ===================================

            # Obtener el indice de la mano mas cercana.
            # El indice permite relacionar landmarks
            # con su lateralidad correspondiente.
            nearest_hand_index = min(
                range(len(result.hand_landmarks)),
                key=lambda index: (
                    sum(
                        point.z
                        for point
                        in result.hand_landmarks[index]
                    )
                    / len(result.hand_landmarks[index])
                )
            )

            nearest_hand = (
                result.hand_landmarks[
                    nearest_hand_index
                ]
            )

            # ===================================
            # OBTENER LATERALIDAD
            # ===================================

            if (
                result.handedness
                and nearest_hand_index
                < len(result.handedness)
                and result.handedness[
                    nearest_hand_index
                ]
            ):

                lateralidad_detectada = (
                result.handedness[
                nearest_hand_index
                ][0].category_name
            )

            # Corregir la lateralidad porque la imagen
            # fue reflejada con cv2.flip(frame, 1)
            if lateralidad_detectada == "Left":
                   lateralidad_actual = "Right"

            elif lateralidad_detectada == "Right":
                lateralidad_actual = "Left"

            else:
                lateralidad_actual = "Unknown"

            alto, ancho, _ = frame.shape

            # Dibujar los 21 puntos
            for landmark in nearest_hand:

                x = int(landmark.x * ancho)
                y = int(landmark.y * alto)

                cv2.circle(
                    frame,
                    (x, y),
                    4,
                    (255, 0, 255),
                    -1
                )

            # Mostrar lateralidad detectada
            cv2.putText(
                frame,
                f"Mano: {lateralidad_actual}",
                (20, 190),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (255, 0, 255),
                2
            )

        tiempo_actual = time.perf_counter()

        # ===================================
        # ESTADO: ACOMODANDO
        # ===================================

        if programa_activo and estado == "ACOMODANDO":

            tiempo_transcurrido = (
                tiempo_actual - inicio_acomodo
            )

            tiempo_restante = max(
                0.0,
                TIEMPO_ACOMODO - tiempo_transcurrido
            )

            cv2.putText(
                frame,
                f"ACOMODATE: {tiempo_restante:.1f} s",
                (20, 45),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.9,
                (0, 255, 255),
                2
            )

            cv2.putText(
                frame,
                "Mueve la mano a otra posicion",
                (20, 80),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                (0, 255, 255),
                2
            )

            # Cuando termina el segundo de acomodo,
            # comienza la captura
            if tiempo_transcurrido >= TIEMPO_ACOMODO:

                estado = "CAPTURANDO"

                secuencia_actual = []
                lateralidades_secuencia = []
                contador_frames = 0

                ultimo_tiempo_captura = (
                    time.perf_counter()
                    - INTERVALO_FRAME
                )

                print(
                    f"Capturando muestra "
                    f"{contador_muestras + 1}/"
                    f"{MAX_MUESTRAS}"
                )

        # ===================================
        # ESTADO: CAPTURANDO
        # ===================================

        if programa_activo and estado == "CAPTURANDO":

            cv2.putText(
                frame,
                f"CAPTURANDO {LETRA} - NO TE MUEVAS",
                (20, 45),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (0, 255, 0),
                2
            )

            cv2.putText(
                frame,
                (
                    f"Muestra: "
                    f"{contador_muestras + 1}/"
                    f"{MAX_MUESTRAS}"
                ),
                (20, 80),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 0),
                2
            )

            cv2.putText(
                frame,
                (
                    f"Frame: "
                    f"{contador_frames}/"
                    f"{FRAMES_POR_MUESTRA}"
                ),
                (20, 115),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 0),
                2
            )

            # Guardar un frame cada 0.05 segundos
            if (
                mano_detectada
                and nearest_hand is not None
                and tiempo_actual - ultimo_tiempo_captura
                >= INTERVALO_FRAME
            ):

                datos_frame = []

                for landmark in nearest_hand:

                    datos_frame.extend([
                        landmark.x,
                        landmark.y,
                        landmark.z
                    ])

                secuencia_actual.extend(
                    datos_frame
                )

                # Guardar la lateralidad de este frame
                lateralidades_secuencia.append(
                    lateralidad_actual
                )

                contador_frames += 1

                ultimo_tiempo_captura = (
                    tiempo_actual
                )

                print(
                    f"Muestra "
                    f"{contador_muestras + 1}/"
                    f"{MAX_MUESTRAS} | "
                    f"Frame "
                    f"{contador_frames}/"
                    f"{FRAMES_POR_MUESTRA} | "
                    f"Mano: {lateralidad_actual}"
                )

                # ===================================
                # GUARDAR MUESTRA COMPLETA
                # ===================================

                if contador_frames == FRAMES_POR_MUESTRA:

                    # Contar cuantas veces se detecto
                    # cada lateralidad en los 20 frames
                    cantidad_right = (
                        lateralidades_secuencia.count(
                            "Right"
                        )
                    )

                    cantidad_left = (
                        lateralidades_secuencia.count(
                            "Left"
                        )
                    )

                    # Seleccionar la lateralidad
                    # detectada en mas frames
                    if cantidad_right > cantidad_left:

                        lateralidad_muestra = "Right"

                    elif cantidad_left > cantidad_right:

                        lateralidad_muestra = "Left"

                    else:

                        lateralidad_muestra = "Unknown"

                    # Estructura final:
                    # 1260 coordenadas
                    # + lateralidad
                    # + letra
                    fila = (
                        secuencia_actual
                        + [lateralidad_muestra]
                        + [LETRA]
                    )

                    writer.writerow(fila)
                    file.flush()

                    contador_muestras += 1

                    print(
                        f"Muestra completa guardada: "
                        f"{contador_muestras}/"
                        f"{MAX_MUESTRAS} | "
                        f"Lateralidad: "
                        f"{lateralidad_muestra} | "
                        f"Right: {cantidad_right} | "
                        f"Left: {cantidad_left}"
                    )

                    # Limpiar la secuencia
                    secuencia_actual = []
                    lateralidades_secuencia = []
                    contador_frames = 0

                    # Dar un segundo para cambiar
                    # posicion antes de la siguiente
                    estado = "ACOMODANDO"

                    inicio_acomodo = (
                        time.perf_counter()
                    )

        # ===================================
        # MENSAJES GENERALES
        # ===================================

        if not programa_activo:

            cv2.putText(
                frame,
                "PAUSADO - ESPACIO para iniciar",
                (20, 45),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.75,
                (0, 255, 255),
                2
            )

            cv2.putText(
                frame,
                (
                    f"Muestras: "
                    f"{contador_muestras}/"
                    f"{MAX_MUESTRAS}"
                ),
                (20, 80),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 255),
                2
            )

        if not mano_detectada:

            cv2.putText(
                frame,
                "MANO NO DETECTADA",
                (20, 155),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 0, 255),
                2
            )

        cv2.imshow(
            "Capturando Dataset Temporal",
            frame
        )

        # ESC = terminar manualmente
        if key == 27:

            print("Captura detenida manualmente.")
            break

# ===================================
# CERRAR RECURSOS
# ===================================

cap.release()
detector.close()
cv2.destroyAllWindows()

print("\nCaptura finalizada.")

print(
    f"Dataset guardado en:\n{CSV_PATH}"
)

print(
    f"Muestras completas guardadas: "
    f"{contador_muestras}"
)

print(
    "Variables por muestra: "
    f"{FRAMES_POR_MUESTRA * 21 * 3}"
)

print(
    "Columnas adicionales: "
    "lateralidad y letra"
)
```

```python
import cv2
import csv
import os
import mediapipe as mp
import time

from mediapipe.tasks import python
from mediapipe.tasks.python import vision

# ===================================
# CONFIGURACION
# ===================================

LETRA = "D"

MAX_MUESTRAS = 300

FRAMES_POR_MUESTRA = 20

INTERVALO_FRAME = 0.05

TIEMPO_ACOMODO = 1.0

MODEL_PATH = (
    r"C:\Proyectos\LSM_Proyecto"
    r"\modelos\hand_landmarker.task"
)

CSV_PATH = rf"C:\Proyectos\LSM_Proyecto\dataset\{LETRA}.csv"

# ===================================
# PREPARACION
# ===================================

os.makedirs(
    os.path.dirname(CSV_PATH),
    exist_ok=True
)

base_options = python.BaseOptions(
    model_asset_path=MODEL_PATH
)

options = vision.HandLandmarkerOptions(
base_options=base_options,
num_hands=1,
min_hand_detection_confidence=0.2,
min_hand_presence_confidence=0.2,
min_tracking_confidence=0.2
)

detector = vision.HandLandmarker.create_from_options(
    options
)

cap = cv2.VideoCapture(0)

# ===================================
# VARIABLES DE CONTROL
# ===================================

contador_muestras = 0
contador_frames = 0

programa_activo = False

estado = "ACOMODANDO"

secuencia_actual = []

inicio_acomodo = 0.0
ultimo_tiempo_captura = 0.0

# ===================================
# CREAR CSV
# ===================================

with open(
    CSV_PATH,
    "w",
    newline=""
) as file:

    writer = csv.writer(file)

    while contador_muestras < MAX_MUESTRAS:

        success, frame = cap.read()

        if not success:
            print("No se pudo obtener imagen de la camara.")
            break

        # Vista espejo
        frame = cv2.flip(frame, 1)

        key = cv2.waitKey(1) & 0xFF

        # ===================================
        # ESPACIO = INICIAR / PAUSAR
        # ===================================

        if key == 32:

            if not programa_activo:

                programa_activo = True

                estado = "ACOMODANDO"

                secuencia_actual = []

                contador_frames = 0

                inicio_acomodo = time.perf_counter()

                print("Programa iniciado.")
                print("Tienes 1 segundo para acomodarte.")

            else:

                programa_activo = False

                estado = "ACOMODANDO"

                secuencia_actual = []

                contador_frames = 0

                print("Captura pausada.")
                print("La muestra incompleta fue descartada.")

        # ===================================
        # PROCESAR IMAGEN
        # ===================================

        rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        mp_image = mp.Image(
            image_format=mp.ImageFormat.SRGB,
            data=rgb
        )

        result = detector.detect(mp_image)

        mano_detectada = False
        nearest_hand = None

        if result.hand_landmarks:

            mano_detectada = True

            nearest_hand = result.hand_landmarks[0]

            alto, ancho, _ = frame.shape

            for landmark in nearest_hand:

                x = int(landmark.x * ancho)
                y = int(landmark.y * alto)

                cv2.circle(
                    frame,
                    (x, y),
                    4,
                    (255, 0, 255),
                    -1
                )

        tiempo_actual = time.perf_counter()

        # ===================================
        # ESTADO ACOMODANDO
        # ===================================

        if programa_activo and estado == "ACOMODANDO":

            tiempo_transcurrido = (
                tiempo_actual - inicio_acomodo
            )

            tiempo_restante = max(
                0.0,
                TIEMPO_ACOMODO - tiempo_transcurrido
            )

            cv2.putText(
                frame,
                f"ACOMODATE: {tiempo_restante:.1f}s",
                (20, 45),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.9,
                (0, 255, 255),
                2
            )

            cv2.putText(
                frame,
                "Mueve la mano a otra posicion",
                (20, 80),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                (0, 255, 255),
                2
            )

            if tiempo_transcurrido >= TIEMPO_ACOMODO:

                estado = "CAPTURANDO"

                secuencia_actual = []

                contador_frames = 0

                ultimo_tiempo_captura = (
                    time.perf_counter()
                    - INTERVALO_FRAME
                )

                print(
                    f"Capturando muestra "
                    f"{contador_muestras + 1}/"
                    f"{MAX_MUESTRAS}"
                )

        # ===================================
        # ESTADO CAPTURANDO
        # ===================================

        if programa_activo and estado == "CAPTURANDO":

            cv2.putText(
                frame,
                f"CAPTURANDO {LETRA}",
                (20, 45),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (0, 255, 0),
                2
            )

            cv2.putText(
                frame,
                (
                    f"Muestra: "
                    f"{contador_muestras + 1}/"
                    f"{MAX_MUESTRAS}"
                ),
                (20, 80),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 0),
                2
            )

            cv2.putText(
                frame,
                (
                    f"Frame: "
                    f"{contador_frames}/"
                    f"{FRAMES_POR_MUESTRA}"
                ),
                (20, 115),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 0),
                2
            )

            if (
                mano_detectada
                and nearest_hand is not None
                and tiempo_actual - ultimo_tiempo_captura
                >= INTERVALO_FRAME
            ):

                datos_frame = []

                for landmark in nearest_hand:

                    datos_frame.extend([
                        landmark.x,
                        landmark.y,
                        landmark.z
                    ])

                secuencia_actual.extend(
                    datos_frame
                )

                contador_frames += 1

                ultimo_tiempo_captura = (
                    tiempo_actual
                )

                print(
                    f"Muestra "
                    f"{contador_muestras + 1}/"
                    f"{MAX_MUESTRAS} | "
                    f"Frame "
                    f"{contador_frames}/"
                    f"{FRAMES_POR_MUESTRA}"
                )

                # ===================================
                # GUARDAR MUESTRA COMPLETA
                # ===================================

                if contador_frames == FRAMES_POR_MUESTRA:

                    fila = (
                        secuencia_actual
                        + [LETRA]
                    )

                    writer.writerow(fila)

                    file.flush()

                    contador_muestras += 1

                    print(
                        f"Muestra completa guardada: "
                        f"{contador_muestras}/"
                        f"{MAX_MUESTRAS}"
                    )

                    secuencia_actual = []

                    contador_frames = 0

                    estado = "ACOMODANDO"

                    inicio_acomodo = (
                        time.perf_counter()
                    )

        # ===================================
        # MENSAJES GENERALES
        # ===================================

        if not programa_activo:

            cv2.putText(
                frame,
                "PAUSADO - ESPACIO para iniciar",
                (20, 45),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.75,
                (0, 255, 255),
                2
            )

            cv2.putText(
                frame,
                (
                    f"Muestras: "
                    f"{contador_muestras}/"
                    f"{MAX_MUESTRAS}"
                ),
                (20, 80),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 255),
                2
            )

        if not mano_detectada:

            cv2.putText(
                frame,
                "MANO NO DETECTADA",
                (20, 155),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 0, 255),
                2
            )

        cv2.imshow(
            "Capturando Dataset Temporal",
            frame
        )

        # ESC = salir

        if key == 27:

            print("Captura detenida manualmente.")

            break

# ===================================
# CERRAR
# ===================================

cap.release()

detector.close()

cv2.destroyAllWindows()

print("\nCaptura finalizada.")

print(
    f"Dataset guardado en:\n{CSV_PATH}"
)

print(
    f"Muestras completas guardadas: "
    f"{contador_muestras}"
)

print(
    "Variables por muestra: "
    f"{FRAMES_POR_MUESTRA * 21 * 3}"
)
```

```python
import cv2
import csv
import os
import mediapipe as mp
import time

from mediapipe.tasks import python
from mediapipe.tasks.python import vision

# ===================================
# CONFIGURACION
# ===================================

LETRA = "J"

MAX_MUESTRAS = 200

FRAMES_POR_MUESTRA = 20

INTERVALO_FRAME = 0.05

TIEMPO_ACOMODO = 1.0

MODEL_PATH = (
    r"C:\Proyectos\LSM_Proyecto"
    r"\modelos\hand_landmarker.task"
)

CSV_PATH = rf"C:\Proyectos\LSM_Proyecto\dataset\{LETRA}.csv"

# ===================================
# PREPARACION
# ===================================

os.makedirs(
    os.path.dirname(CSV_PATH),
    exist_ok=True
)

base_options = python.BaseOptions(
    model_asset_path=MODEL_PATH
)

options = vision.HandLandmarkerOptions(
base_options=base_options,
running_mode=vision.RunningMode.VIDEO,
num_hands=1,
min_hand_detection_confidence=0.3,
min_hand_presence_confidence=0.2,
min_tracking_confidence=0.2
)

detector = vision.HandLandmarker.create_from_options(
    options
)

cap = cv2.VideoCapture(0)

# ===================================
# VARIABLES DE CONTROL
# ===================================

contador_muestras = 0
contador_frames = 0

programa_activo = False

estado = "ACOMODANDO"

secuencia_actual = []

inicio_acomodo = 0.0
ultimo_tiempo_captura = 0.0
frame_number = 0

# ===================================
# CREAR CSV
# ===================================

with open(
    CSV_PATH,
    "w",
    newline=""
) as file:

    writer = csv.writer(file)

    while contador_muestras < MAX_MUESTRAS:

        success, frame = cap.read()

        if not success:
            print("No se pudo obtener imagen de la camara.")
            break

        # Vista espejo
        frame = cv2.flip(frame, 1)

        key = cv2.waitKey(1) & 0xFF

        # ===================================
        # ESPACIO = INICIAR / PAUSAR
        # ===================================

        if key == 32:

            if not programa_activo:

                programa_activo = True

                estado = "ACOMODANDO"

                secuencia_actual = []

                contador_frames = 0

                inicio_acomodo = time.perf_counter()

                print("Programa iniciado.")
                print("Tienes 1 segundo para acomodarte.")

            else:

                programa_activo = False

                estado = "ACOMODANDO"

                secuencia_actual = []

                contador_frames = 0

                print("Captura pausada.")
                print("La muestra incompleta fue descartada.")

        # ===================================
        # PROCESAR IMAGEN
        # ===================================

        rgb = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2RGB
        )

        mp_image = mp.Image(
            image_format=mp.ImageFormat.SRGB,
            data=rgb
        )

        frame_number += 1

        timestamp_ms = frame_number * 33

        result = detector.detect_for_video(
            mp_image,
            timestamp_ms
        )

        mano_detectada = False
        nearest_hand = None

        if result.hand_landmarks:

            mano_detectada = True

            nearest_hand = result.hand_landmarks[0]

            alto, ancho, _ = frame.shape

            for landmark in nearest_hand:

                x = int(landmark.x * ancho)
                y = int(landmark.y * alto)

                cv2.circle(
                    frame,
                    (x, y),
                    4,
                    (255, 0, 255),
                    -1
                )

        tiempo_actual = time.perf_counter()

        # ===================================
        # ESTADO ACOMODANDO
        # ===================================

        if programa_activo and estado == "ACOMODANDO":

            tiempo_transcurrido = (
                tiempo_actual - inicio_acomodo
            )

            tiempo_restante = max(
                0.0,
                TIEMPO_ACOMODO - tiempo_transcurrido
            )

            cv2.putText(
                frame,
                f"ACOMODATE: {tiempo_restante:.1f}s",
                (20, 45),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.9,
                (0, 255, 255),
                2
            )

            cv2.putText(
                frame,
                "Mueve la mano a otra posicion",
                (20, 80),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.65,
                (0, 255, 255),
                2
            )

            if tiempo_transcurrido >= TIEMPO_ACOMODO:

                estado = "CAPTURANDO"

                secuencia_actual = []

                contador_frames = 0

                ultimo_tiempo_captura = (
                    time.perf_counter()
                    - INTERVALO_FRAME
                )

                print(
                    f"Capturando muestra "
                    f"{contador_muestras + 1}/"
                    f"{MAX_MUESTRAS}"
                )

        # ===================================
        # ESTADO CAPTURANDO
        # ===================================

        if programa_activo and estado == "CAPTURANDO":

            cv2.putText(
                frame,
                f"CAPTURANDO {LETRA}",
                (20, 45),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (0, 255, 0),
                2
            )

            cv2.putText(
                frame,
                (
                    f"Muestra: "
                    f"{contador_muestras + 1}/"
                    f"{MAX_MUESTRAS}"
                ),
                (20, 80),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 0),
                2
            )

            cv2.putText(
                frame,
                (
                    f"Frame: "
                    f"{contador_frames}/"
                    f"{FRAMES_POR_MUESTRA}"
                ),
                (20, 115),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 0),
                2
            )

            if (
                mano_detectada
                and nearest_hand is not None
                and tiempo_actual - ultimo_tiempo_captura
                >= INTERVALO_FRAME
            ):

                datos_frame = []

                for landmark in nearest_hand:

                    datos_frame.extend([
                        landmark.x,
                        landmark.y,
                        landmark.z
                    ])

                secuencia_actual.extend(
                    datos_frame
                )

                contador_frames += 1

                ultimo_tiempo_captura = (
                    tiempo_actual
                )

                print(
                    f"Muestra "
                    f"{contador_muestras + 1}/"
                    f"{MAX_MUESTRAS} | "
                    f"Frame "
                    f"{contador_frames}/"
                    f"{FRAMES_POR_MUESTRA}"
                )

                # ===================================
                # GUARDAR MUESTRA COMPLETA
                # ===================================

                if contador_frames == FRAMES_POR_MUESTRA:

                    fila = (
                        secuencia_actual
                        + [LETRA]
                    )

                    writer.writerow(fila)

                    file.flush()

                    contador_muestras += 1

                    print(
                        f"Muestra completa guardada: "
                        f"{contador_muestras}/"
                        f"{MAX_MUESTRAS}"
                    )

                    secuencia_actual = []

                    contador_frames = 0

                    estado = "ACOMODANDO"

                    inicio_acomodo = (
                        time.perf_counter()
                    )

        # ===================================
        # MENSAJES GENERALES
        # ===================================

        if not programa_activo:

            cv2.putText(
                frame,
                "PAUSADO - ESPACIO para iniciar",
                (20, 45),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.75,
                (0, 255, 255),
                2
            )

            cv2.putText(
                frame,
                (
                    f"Muestras: "
                    f"{contador_muestras}/"
                    f"{MAX_MUESTRAS}"
                ),
                (20, 80),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 255, 255),
                2
            )

        if not mano_detectada:

            cv2.putText(
                frame,
                "MANO NO DETECTADA",
                (20, 155),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.7,
                (0, 0, 255),
                2
            )

        cv2.imshow(
            "Capturando Dataset Temporal",
            frame
        )

        # ESC = salir

        if key == 27:

            print("Captura detenida manualmente.")

            break

# ===================================
# CERRAR
# ===================================

cap.release()

detector.close()

cv2.destroyAllWindows()

print("\nCaptura finalizada.")

print(
    f"Dataset guardado en:\n{CSV_PATH}"
)

print(
    f"Muestras completas guardadas: "
    f"{contador_muestras}"
)

print(
    "Variables por muestra: "
    f"{FRAMES_POR_MUESTRA * 21 * 3}"
)
```
